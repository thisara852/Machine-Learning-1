#### What is Machine Learning?

Machine Learning is a field of Artificial Intelligence where computers learn from data and improve their performance over time without being explicitly programmed for every situation.

#### Name the three main types of Machine Learning.

Supervised Learning
Unsupervised Learning
Reinforcement Learning

#### What is Supervised Learning?

Supervised Learning is a type of machine learning where the model is trained using labeled data, meaning each input already has the correct output. The system learns from these examples and uses them to make predictions.

#### What is Unsupervised Learning?

Unsupervised Learning is a type of machine learning where the system is given data without labels. It tries to find hidden patterns or naturally group similar data together.

#### What is Reinforcement Learning?

Reinforcement Learning is a type of machine learning where an agent learns by interacting with its environment. It improves by receiving rewards for correct actions and penalties for incorrect actions.
